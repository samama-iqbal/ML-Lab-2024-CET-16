Fix the problems **in this order** (text first, then duplicates):

```python
# Task 1: Load and first look

import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\asif\Downloads\pakistan_used_cars.csv")

# (a) Shape
print("Shape of dataset:")
print(df.shape)

# (b) First 5 rows
print("\nFirst 5 rows:")
print(df.head())

# (c) Column names
print("\nColumn names:")
print(df.columns.tolist())

# (d) Data types
print("\nData types:")
print(df.dtypes)

# (e) Price column examples
print("\nPrice examples:")
print(df["Price"].head(10).tolist())
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# T4: Analyse

import pandas as pd
import numpy as np

# Make a copy of original dataset
df = df.copy()

# Clean column names
df.columns = df.columns.str.strip()

# Convert Price into numeric values
df["Price"] = pd.to_numeric(
    df["Price"].astype(str).str.replace(",", "").str.strip(),
    errors="coerce"
)

# (a) Price in lakh rupees
df["Price_lakh"] = df["Price"] / 100000

print("T4(a) - Price in Lakh:")
print(df[["Price", "Price_lakh"]].head())


# (b) Calculate Age
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df["Age"] = 2020 - df["Year"]

print("\nT4(b) - Car Age:")
print(df[["Year", "Age"]].head())


# (c) 5 Most Common Brands
brand_col = next(
    (c for c in ["Brand", "Make", "Company"] if c in df.columns),
    None
)

if brand_col:
    print("\nT4(c) - Top 5 Most Common Brands:")
    print(df[brand_col].value_counts().head(5))
else:
    print("Brand column not found. Available columns:", df.columns.tolist())


# (d) 5 Cities with Most Ads
city_col = next(
    (c for c in ["City", "city"] if c in df.columns),
    None
)

if city_col:
    print("\nT4(d) - Top 5 Cities:")
    print(df[city_col].value_counts().head(5))
else:
    print("City column not found.")


# (e) Median Price for Each Fuel Type
fuel_col = next(
    (c for c in ["Fuel", "Fuel_Type", "Fuel Type"] if c in df.columns),
    None
)

if fuel_col:
    print("\nT4(e) - Median Price by Fuel Type:")
    print(
        df.groupby(fuel_col)["Price_lakh"]
        .median()
        .sort_values(ascending=False)
    )
else:
    print("Fuel column not found.")


# (f) Corolla GLI Median Price for Each Year (2010-2018)

model_col = next(
    (c for c in ["Model", "model"] if c in df.columns),
    None
)

if model_col:
    corolla = df[
        df[model_col].astype(str).str.contains(
            "Corolla GLI", case=False, na=False
        )
        & df["Year"].between(2010, 2018)
    ]

    yearly_price = corolla.groupby("Year")["Price_lakh"].median()

    print("\nT4(f) - Corolla GLI Median Price:")
    print(yearly_price)

    if len(yearly_price.dropna()) >= 2:
        first_year = yearly_price.dropna().index.min()
        last_year = yearly_price.dropna().index.max()

        value_loss = (
            yearly_price.loc[first_year] -
            yearly_price.loc[last_year]
        ) / (last_year - first_year)

        print("Average value loss per year:", value_loss, "Lakh")
    else:
        print("Not enough data to calculate yearly value loss.")
else:
    print("Model column not found.")


# (g) 5 Most Expensive Ads
print("\nT4(g) - Top 5 Most Expensive Ads:")
print(
    df.sort_values("Price", ascending=False).head(5)
)

print("\nCheck these ads for unusually high or incorrect prices.")


# (h) Cash vs Installment/Leasing

payment_col = next(
    (c for c in ["Payment Type", "Payment_Type",
                 "Payment", "payment_type"] if c in df.columns),
    None
)

if payment_col:
    print("\nT4(h) - Number of Ads:")
    print(df[payment_col].value_counts())

    print("\nMedian Price by Payment Type:")
    print(
        df.groupby(payment_col)["Price_lakh"]
        .median()
        .sort_values(ascending=False)
    )
else:
    print("\nPayment Type column not found.")
    print("Available columns:", df.columns.tolist())
```
