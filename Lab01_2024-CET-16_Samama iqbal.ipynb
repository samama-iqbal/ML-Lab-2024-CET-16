Function: Grab the domain from "user@domain.com".

```python
def get_domain(email):
    return email.split('@')[1]

print(get_domain("user@domain.com"))
```

TASK 08

Function: return True if the word "dog" is in a string(case_insensitive).

```python
def find_dog(text):
    return "dog" in text.lower()

print(find_dog("Is there a dog here?"))
```

TASK 09

Function:count how many times "dog" occurs in string(casr-insensitive).

```python
def count_dog(text):
    return text.lower().count("dog")

print(count_dog("This dog runs faster than the other dog dude!"))
```
