# Task 3: Randomness
A. Set seed 2026 and generate one random value.
B. Generate 25 random values.
C. Set seed 42 and roll 10,000 Ludo dice.
D. Count each face using np.unique.
E. Estimate P(six) and count double sixes without a loop.

```python
# A: One random value
np.random.seed(2026)
one_random = np.random.rand()
print("T3A - One Random Value:", one_random)

# B: 25 random values
random_values = np.random.rand(25)
print("T3B - 25 Random Values:\n", random_values)

# C: Roll two dice 10,000 times
np.random.seed(42)
dice_rolls = np.random.randint(1, 7, size=(10000, 2))
print("T3C - Dice Rolls Shape:", dice_rolls.shape)

# D: Count each face
all_faces, face_counts = np.unique(
    dice_rolls, return_counts=True
)
print("T3D - Dice Faces:", all_faces)
print("T3D - Face Counts:", face_counts)

# E: Probability of getting six
total_sixes = np.sum(dice_rolls == 6)
probability_six = total_sixes / dice_rolls.size

double_sixes = np.sum(
    np.all(dice_rolls == 6, axis=1)
)

print("T3E - Estimated P(Six):", probability_six)
print("T3E - Double Sixes:", double_sixes)
```

# Task 6: Census 2023
A. Calculate the total population.
B. Calculate the percentage share of each unit.
C. Find the most populous unit.
D. Count units above 20 million.
E. Calculate annual growth from 2017 to 2023.
F. Find the fastest-growing unit, interpret growth and compare AJK.

```python
# Sample Pakistan Census 2023 population data (millions)
units = np.array([
    "Punjab", "Sindh", "Khyber Pakhtunkhwa",
    "Balochistan", "Islamabad", "AJK", "Gilgit-Baltistan"
])

population = np.array([
    127.69, 55.70, 40.85, 14.89, 2.36, 4.46, 1.71
])

# A: Total population
total_population = np.sum(population)
print("T6A - Total Population (millions):", total_population)

# B: Percentage share
percentage_share = (population / total_population) * 100

print("T6B - Population Share (%):")
print(np.round(percentage_share, 2))

# C: Most populous unit
most_populous = units[np.argmax(population)]
print("T6C - Most Populous Unit:", most_populous)

# D: Units above 20 million
above_20 = np.sum(population > 20)
print("T6D - Units Above 20 Million:", above_20)

# E: Sample 2017 population estimates (millions)
population_2017 = np.array([
    110.01, 47.85, 35.50, 12.34, 2.00, 4.05, 1.49
])

annual_growth = (
    (population / population_2017) ** (1 / 6) - 1
) * 100

print("T6E - Annual Growth Rate (%):")
print(np.round(annual_growth, 2))

# F: Fastest-growing unit
fastest_index = np.argmax(annual_growth)

print("T6F - Fastest Growing Unit:", units[fastest_index])
print("Growth Rate (%):", round(annual_growth[fastest_index], 2))

# AJK comparison
ajk_index = np.where(units == "AJK")[0][0]

print("AJK Population (millions):", population[ajk_index])
print("AJK Annual Growth (%):", round(annual_growth[ajk_index], 2))

print(
    "Interpretation: Population growth differs across administrative "
    "units. Punjab has the largest population, while smaller units "
    "have different growth rates."
)
```

# Task 7: Climate
A. Display the shape of climate data.
B. Calculate yearly mean temperature per city.
C. Find the hottest month per city using argmax.
D. Calculate the five-city monthly average.
E. Count months above 35°C.
F. Convert Lahore temperature into Fahrenheit.
G. Calculate monsoon rainfall from July to September.
H. Find the wettest monsoon city and its monsoon rainfall share.
Interpret one surprising result.

```python
# Sample temperature data for five cities (12 months)
cities = np.array([
    "Lahore", "Karachi", "Islamabad", "Multan", "Peshawar"
])

months = np.array([
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
])

temperature = np.array([
    [13, 17, 23, 30, 36, 38, 34, 33, 32, 28, 21, 15],
    [20, 22, 25, 28, 31, 32, 30, 29, 30, 29, 25, 21],
    [9, 12, 18, 25, 31, 35, 30, 29, 28, 24, 17, 11],
    [15, 19, 26, 33, 39, 42, 37, 36, 34, 30, 23, 17],
    [10, 14, 21, 28, 35, 38, 33, 32, 31, 26, 18, 12]
])

# Sample monthly monsoon rainfall (mm)
rainfall = np.array([
    [20, 25, 40, 30, 20, 50, 180, 200, 120, 25, 10, 5],
    [5, 5, 5, 5, 10, 20, 60, 70, 50, 10, 5, 5],
    [30, 35, 50, 40, 30, 70, 250, 280, 160, 30, 15, 10],
    [10, 15, 20, 15, 10, 25, 100, 120, 80, 15, 5, 5],
    [15, 20, 30, 25, 15, 40, 150, 170, 100, 20, 10, 5]
])

# A: Shape
print("T7A - Temperature Shape:", temperature.shape)

# B: Yearly mean per city
yearly_mean = np.mean(temperature, axis=1)
print("T7B - Yearly Mean Temperature (°C):")
print(np.round(yearly_mean, 2))

# C: Hottest month per city
hottest_month_index = np.argmax(temperature, axis=1)

print("T7C - Hottest Month:")
print(months[hottest_month_index])

# D: Five-city monthly average
monthly_average = np.mean(temperature, axis=0)
print("T7D - Monthly Average Temperature (°C):")
print(np.round(monthly_average, 2))

# E: Months above 35°C per city
above_35 = np.sum(temperature > 35, axis=1)
print("T7E - Months Above 35°C:")
print(above_35)

# F: Lahore Celsius to Fahrenheit
lahore_celsius = temperature[0]
lahore_fahrenheit = (lahore_celsius * 9/5) + 32

print("T7F - Lahore Temperature (°F):")
print(np.round(lahore_fahrenheit, 2))

# G: July to September rainfall
monsoon_rainfall = rainfall[:, 6:9]
monsoon_total = np.sum(monsoon_rainfall, axis=1)

print("T7G - Monsoon Rainfall (mm):")
print(monsoon_total)

# H: Wettest monsoon city
wettest_index = np.argmax(monsoon_total)
total_rainfall = np.sum(rainfall, axis=1)

monsoon_share = (monsoon_total / total_rainfall) * 100

print("T7H - Wettest Monsoon City:", cities[wettest_index])
print("Monsoon Rainfall Share (%):", np.round(monsoon_share, 2))

print(
    "Interpretation: Islamabad receives substantial monsoon rainfall, "
    "while Multan experiences higher summer temperatures."
)
```

# Task 8: Inflation
A. Calculate the average inflation per year.
B. Calculate FY2024 monthly inflation drop.
C. Find the month of biggest drop.
D. Classify FY2025 months as Low or Moderate using np.where.
E. Interpret what happened to prices.

```python
# Sample monthly inflation rates (%)
months = np.array([
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
    "Jan", "Feb", "Mar", "Apr", "May", "Jun"
])

fy2024 = np.array([
    28.3, 27.4, 31.4, 26.9, 29.2, 29.7,
    28.3, 23.1, 20.7, 17.3, 11.8, 12.6
])

fy2025 = np.array([
    11.1, 9.6, 6.9, 7.2, 4.9, 4.1,
    2.4, 1.5, 0.7, 0.3, 3.5, 3.2
])

# A: Average inflation per year
average_2024 = np.mean(fy2024)
average_2025 = np.mean(fy2025)

print("T8A - FY2024 Average Inflation (%):", round(average_2024, 2))
print("T8A - FY2025 Average Inflation (%):", round(average_2025, 2))

# B: Monthly inflation drop
monthly_drop = fy2024 - fy2025
print("T8B - Monthly Inflation Drop (%):")
print(np.round(monthly_drop, 2))

# C: Month of biggest drop
biggest_drop_index = np.argmax(monthly_drop)

print("T8C - Biggest Drop Month:", months[biggest_drop_index])
print("Biggest Drop (%):", round(monthly_drop[biggest_drop_index], 2))

# D: Classify FY2025 months
classification = np.where(
    fy2025 < 5,
    "Low",
    "Moderate"
)

print("T8D - FY2025 Inflation Classification:")
print(classification)

# E: Interpretation
print(
    "T8E - Interpretation: Inflation rates decreased in FY2025 "
    "compared with FY2024. However, lower inflation does not mean "
    "that prices have returned to their previous levels."
)
```

# Task 9: Grace Marks
A. Add 5 grace marks to each student.
B. Cap marks at 100 using np.clip.
C. Find students who passed with 96 marks.
D. Calculate z-scores.

```python
# Original marks of 10 students
marks = np.array([45, 55, 62, 70, 88, 91, 94, 96, 98, 99])

# A: Add 5 grace marks
grace_marks = marks + 5
print("T9A - Marks After Grace:", grace_marks)

# B: Cap marks at 100
final_marks = np.clip(grace_marks, 0, 100)
print("T9B - Final Marks:", final_marks)

# C: Students who passed only because of grace marks
passed_with_grace = (marks < 50) & (final_marks >= 50)

print("T9C - Passed Only Because of Grace:")
print(passed_with_grace)

print("Number of Students:", np.sum(passed_with_grace))

# D: Calculate z-scores
mean_marks = np.mean(final_marks)
std_marks = np.std(final_marks)

if std_marks != 0:
    z_scores = (final_marks - mean_marks) / std_marks
else:
    z_scores = np.zeros_like(final_marks, dtype=float)

print("T9D - Mean Marks:", mean_marks)
print("T9D - Standard Deviation:", std_marks)
print("T9D - Z-Scores:", np.round(z_scores, 2))
```

# Task 10: Image Is an Array
A. Draw the number 7 on an 8 × 8 canvas.
B. Flip the image.
C. Invert the image.
D. Crop the image.
E. Brighten the image with a cap.
F. Display shape, dtype and white pixels.

```python
# A: Create an 8 × 8 white canvas
image = np.zeros((8, 8), dtype=np.uint8)

# Draw number 7
image[0, :] = 255
image[:, 7] = 255
image[1, 6] = 255
image[2, 5] = 255
image[3, 4] = 255
image[4, 3] = 255
image[5, 2] = 255
image[6, 1] = 255
image[7, 0] = 255

print("T10A - Original Image:\n", image)

# B: Flip the image horizontally
flipped_image = np.fliplr(image)
print("T10B - Flipped Image:\n", flipped_image)

# C: Invert the image
inverted_image = 255 - image
print("T10C - Inverted Image:\n", inverted_image)

# D: Crop the image
cropped_image = image[0:6, 1:7]
print("T10D - Cropped Image:\n", cropped_image)

# E: Brighten the image and cap at 255
brightened_image = np.clip(
    image.astype(np.int16) + 50,
    0,
    255
).astype(np.uint8)

print("T10E - Brightened Image:\n", brightened_image)

# F: Shape, dtype and white pixels
print("T10F - Shape:", image.shape)
print("T10F - Data Type:", image.dtype)
print("T10F - White Pixels:", np.sum(image == 255))
```
